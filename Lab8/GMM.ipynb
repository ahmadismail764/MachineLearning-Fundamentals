# Clustering 2: Gaussian Mixture Models (GMM)

**Dataset:** Iris (Features only)

**Objective:** Group data using a probabilistic approach. Unlike K-Means, GMM accounts for cluster covariance and provides soft assignments.

**Key Concepts:**

- **Probabilistic:** GMM assumes data is generated from a mixture of finite Gaussian distributions.
- **Soft Clustering:** Points belong to a cluster with a certain probability, rather than 0 or 1.
- **Covariance Type:** Determines the shape of the clusters (spherical, tied, diag, or full).

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
X = df.values
```

### Step 2: Apply GMM

We fit a GMM with 3 components.


```python
gmm = GaussianMixture(n_components=3, random_state=42)
labels = gmm.fit_predict(X)

score = silhouette_score(X, labels)
print(f'Silhouette Score for GMM (3 components): {score:.4f}')

df['Cluster'] = labels
```

### Step 3: Probability of Membership

We can check the probability that a specific point belongs to each cluster.


```python
probs = gmm.predict_proba(X)
print("Probabilities for the first 5 samples:")
print(probs[:5].round(3))
```

### Step 4: Visualize Clusters


```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal length (cm)', y='petal width (cm)', hue='Cluster', palette='Set1')
plt.title('GMM Clustering on Iris')
plt.show()
```
