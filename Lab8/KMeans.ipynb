# Clustering 1: K-Means

**Dataset:** Iris (Features only)

**Objective:** Use K-Means to group data into clusters. We will use the **Elbow Method** and **Silhouette Score** to find the optimal number of clusters (K).

**Key Concepts:**

- **Centroid-based:** K-Means minimizes the distance between points and their cluster center.
- **WSS (Inertia):** Within-cluster Sum of Squares. Used for the Elbow method.
- **Silhouette Score:** Measures how similar an object is to its own cluster compared to other clusters.

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Load Iris dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

print(f"Dataset Shape: {df.shape}")
df.head()
```

### Step 2: Find Optimal K (Elbow Method)

We plot Inertia vs. number of clusters. The "elbow" point is where adding more clusters gives diminishing returns.


```python
wss = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(df)
    wss.append(kmeans.inertia_)

plt.plot(range(1, 11), wss, marker='o')
plt.title('The Elbow Method')
plt.xlabel('Number of clusters')
plt.ylabel('WSS (Inertia)')
plt.show()
```

### Step 3: Apply K-Means & Silhouette Score

We fit K-Means with K=3 (expected for Iris) and calculate the Silhouette score.


```python
kmeans = KMeans(n_clusters=3, init='k-means++', random_state=42)
labels = kmeans.fit_predict(df)

score = silhouette_score(df, labels)
print(f'Silhouette Score for K=3: {score:.4f}')

# Add labels to dataframe for visualization
df['Cluster'] = labels
```

### Step 4: Visualize Clusters


```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal length (cm)', y='petal width (cm)', hue='Cluster', palette='viridis')
plt.title('K-Means Clustering on Iris (Petal Features)')
plt.show()
```
