# Clustering 3: DBSCAN (Density-Based Spatial Clustering)

**Dataset:** Iris (Features only)

**Objective:** Group data based on density. DBSCAN is great for finding clusters of arbitrary shapes and identifying **outliers** (noise).

**Key Concepts:**

- **Density-based:** Groups together points that are close to each other based on a distance measure (eps) and a minimum number of points (min_samples).
- **Epsilon (eps):** The maximum distance between two samples for one to be considered as in the neighborhood of the other.
- **Noise:** Points that are in low-density areas are labeled as -1.

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.datasets import load_iris
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score

iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# DBSCAN is sensitive to scale, so we standardize first
X = StandardScaler().fit_transform(df.values)
```

### Step 2: Apply DBSCAN

We use `eps=0.5` and `min_samples=5`.


```python
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(X)

n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = list(labels).count(-1)

print(f'Estimated number of clusters: {n_clusters}')
print(f'Estimated number of noise points: {n_noise}')

if n_clusters > 1:
    score = silhouette_score(X, labels)
    print(f'Silhouette Score: {score:.4f}')

df['Cluster'] = labels
```

### Step 3: Visualize Clusters & Noise

Noise points are labeled as -1 (usually black/gray in plots).


```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=df, x='petal length (cm)', y='petal width (cm)', hue='Cluster', palette='Set2')
plt.title('DBSCAN Clustering on Iris (Standardized Features)')
plt.show()
```
