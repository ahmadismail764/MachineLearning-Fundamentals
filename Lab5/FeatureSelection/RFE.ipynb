# Feature Selection 6: Recursive Feature Elimination (RFE)

**Dataset:** `clean_scaled_loan_data.csv`

**When to use:** A greedy optimization algorithm that fits a model, ranks features by importance, and **removes the weakest** features iteratively.

**How it works:**

1. Fit a model (e.g., ExtraTreesRegressor) on all features
2. Rank features by the model's `feature_importances_`
3. Remove the weakest feature(s)
4. Repeat until `n_features_to_select` is reached

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_selection import RFE
from sklearn.ensemble import ExtraTreesRegressor

df = pd.read_csv('../../data/Lab5/clean_scaled_loan_data.csv')

y = df['loan_amount']
X = df.drop(['loan_amount'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100, test_size=0.3)

print(f'Features: {len(X.columns)}, Train: {X_train.shape}, Test: {X_test.shape}')
df.head()
```

### Step 2: Apply RFE

Uses `ExtraTreesRegressor` because the target 'loan_amount' is continuous.


```python
df_rfe = X_train.copy()
print(f'Starting with {len(df_rfe.columns)} features')

reg = ExtraTreesRegressor(n_estimators=60, n_jobs=-1, random_state=42)

n_features_to_select = 10
rfe = RFE(reg, n_features_to_select=n_features_to_select)
rfe.fit(df_rfe, y_train)
```

### Step 3: View Results & Rankings


```python
# Feature rankings (1 = selected)
ranking = pd.Series(rfe.ranking_, index=df_rfe.columns).sort_values()
print('Feature Rankings (1 = selected):')
print(ranking.head(15))

selected = df_rfe.columns[rfe.support_].tolist()
print(f'\nSelected {len(selected)} features: {selected}')
```
