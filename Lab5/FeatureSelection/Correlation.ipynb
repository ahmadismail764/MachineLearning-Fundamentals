# Feature Selection 2: Correlation Coefficient

**Dataset:** `clean_scaled_loan_data.csv`

**When to use:** To find features that have a **high correlation with the target** but **low correlation with each other** (to avoid redundancy/multicollinearity).

**Key concept:** Features with inter-correlation > 0.7 are considered redundant — one of them can be removed.

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv('../../data/Lab5/clean_scaled_loan_data.csv')

y = df['loan_amount']
X = df.drop(['loan_amount'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100, test_size=0.3)

print(f'Features: {len(X.columns)}, Train: {X_train.shape}, Test: {X_test.shape}')
df.head()
```

### Step 2: Compute & Visualize Correlation Matrix


```python
df_corr = X_train.copy()

cor = df_corr.corr()
plt.figure(figsize=(20, 10))
sns.heatmap(cor, cmap=plt.cm.CMRmap_r, annot=True)
plt.title('Feature Correlation Heatmap')
plt.show()
```

### Step 3: Find & Remove Highly Correlated Features

Remove any feature if it has correlation with another feature > 0.7.


```python
def correlation(dataset, threshold):
    col_corr = set()
    corr_matrix = dataset.corr()
    for i in range(len(corr_matrix.columns)):
        for j in range(i):
            if abs(corr_matrix.iloc[i, j]) > threshold:
                colname = corr_matrix.columns[i]
                col_corr.add(colname)
    return col_corr

corr_features = correlation(df_corr, 0.7)
print(f'Features to remove (corr > 0.7): {corr_features}')
```

### Step 4: Apply Feature Removal


```python
df_corr = df_corr.drop(corr_features, axis=1)
print(f'Reduced from {len(X_train.columns)} to {len(df_corr.columns)} features')
df_corr.head()
```
