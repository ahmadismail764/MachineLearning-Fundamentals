# Feature Selection 3: Variance Threshold

**Dataset:** `clean_scaled_loan_data.csv`

**When to use:** A simple baseline approach that removes all features whose **variance doesn't meet a threshold**.

**Key concept:** Features with very low variance (e.g., the same value for 99% of observations) contain very little information and can be safely removed.

---

### Step 1: Setup & Data Loading

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv('../../data/Lab5/clean_scaled_loan_data.csv')

y = df['loan_amount']
X = df.drop(['loan_amount'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100, test_size=0.3)

print(f'Features: {len(X.columns)}, Train: {X_train.shape}, Test: {X_test.shape}')
df.head()
```

### Step 2: Define Variance Threshold Function

```python
from sklearn.feature_selection import VarianceThreshold

def variance_threshold(df, th):
    var_thres = VarianceThreshold(threshold=th)
    var_thres.fit(df)
    new_cols = var_thres.get_support()
    return df.iloc[:, new_cols]

print(f'Features before: {len(X_train.columns)}')
```

### Step 3: Apply Variance Threshold

```python
df_var = X_train.copy()
df_var = variance_threshold(df_var, 0.001)

print(f'Features after (threshold=0.001): {len(df_var.columns)}')
print(f'Remaining features: {df_var.columns.tolist()}')
```
