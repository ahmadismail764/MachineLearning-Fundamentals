# Feature Selection 1: Information Gain (Mutual Information)

**Dataset:** `clean_scaled_loan_data.csv`

**When to use:** To measure how much "information" a feature provides about the target variable. Based on the concept of **Entropy** (impurity measure).

**Key concept:** A higher Information Gain indicates the feature is highly effective at reducing uncertainty about the class.

> **Note:** `mutual_info_classif` is for classification targets (discrete). For regression targets (continuous), use `mutual_info_regression`.

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split

df = pd.read_csv('../../data/Lab5/clean_scaled_loan_data.csv')

y = df['loan_amount']
X = df.drop(['loan_amount'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100, test_size=0.3)

print(f'Features: {len(X.columns)}, Train: {X_train.shape}, Test: {X_test.shape}')
df.head()
```

### Step 2: Calculate Mutual Information

We use `mutual_info_regression` since `loan_amount` is a continuous target variable.


```python
from sklearn.feature_selection import mutual_info_regression

df_ig = X_train.copy()

# Calculate mutual information for regression target
mutual_info = mutual_info_regression(df_ig, y_train)
mutual_info = pd.Series(mutual_info, index=df_ig.columns)
mutual_info.sort_values(ascending=False)
```

### Step 3: Visualize Feature Importance


```python
mutual_info = mutual_info.sort_values(ascending=False)
mutual_info.plot.bar(figsize=(20, 8))
plt.title('Feature Importance by Mutual Information')
plt.ylabel('Mutual Information Score')
plt.tight_layout()
plt.show()
```

### Step 4: Select Top Features


```python
# Select top N features
top_n = 10
top_features = mutual_info.nlargest(top_n).index.tolist()
print(f'Top {top_n} features: {top_features}')
print(f'\nReduced from {len(X_train.columns)} to {top_n} features')
```
