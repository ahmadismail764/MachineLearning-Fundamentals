# Feature Selection 5: Backward Feature Elimination

**Dataset:** `clean_scaled_loan_data.csv`

**When to use:** The opposite of forward selection. Starts with **all features** and removes the least significant one at each step.

**How it works:**

1. Train a model with all features
2. Remove the feature with the smallest impact on performance
3. Repeat until `k_features` is reached

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from mlxtend.feature_selection import SequentialFeatureSelector as sfs
from sklearn.ensemble import ExtraTreesRegressor

df = pd.read_csv('../../data/Lab5/clean_scaled_loan_data.csv')

y = df['loan_amount']
X = df.drop(['loan_amount'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100, test_size=0.3)

print(f'Features: {len(X.columns)}, Train: {X_train.shape}, Test: {X_test.shape}')
df.head()
```

### Step 2: Apply Backward Elimination

Uses `ExtraTreesRegressor` and `scoring='r2'` because the target is continuous.


```python
df_bwd = X_train.copy()
print(f'Starting with {len(df_bwd.columns)} features')

# Use Regressor for continuous target 'loan_amount'
reg = ExtraTreesRegressor(n_estimators=60, n_jobs=-1, random_state=42)

sfs2 = sfs(reg,
           k_features=10,
           forward=False,
           floating=False,
           verbose=2,
           scoring='r2',
           cv=5) # Reduced CV for speed

sfs2.fit(df_bwd, y_train)
```

### Step 3: View Selected Features


```python
print(f'Selected features ({len(sfs2.k_feature_names_)}): {sfs2.k_feature_names_}')

df_bwd_selected = df_bwd[list(sfs2.k_feature_names_)]
df_bwd_selected.head()
```
