# Feature Selection 4: Forward Feature Selection

**Dataset:** `clean_scaled_loan_data.csv`

**When to use:** An iterative **wrapper method** that starts with **no features** and adds the best-performing feature one at a time.

**How it works:**

1. Start with an empty model
2. Try each feature individually, keep the best one
3. Try combinations (best + each remaining), keep the best pair
4. Repeat until `k_features` is reached

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from mlxtend.feature_selection import SequentialFeatureSelector as sfs
from sklearn.ensemble import ExtraTreesRegressor

df = pd.read_csv('../../data/Lab5/clean_scaled_loan_data.csv')

y = df['loan_amount']
X = df.drop(['loan_amount'], axis=1)
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=100, test_size=0.3)

print(f'Features: {len(X.columns)}, Train: {X_train.shape}, Test: {X_test.shape}')
df.head()
```

### Step 2: Apply Forward Selection

Uses `ExtraTreesRegressor` and `scoring='r2'`. We use a subset of features to speed up the demonstration.


```python
df_fwd = X_train.copy()
subset_cols = ["income", "submission_of_application_freq", "loan_limit_freq", "Security_Type_freq", "age"]
df_fwd_subset = df_fwd[subset_cols]

print(f'Starting with {len(df_fwd_subset.columns)} features: {df_fwd_subset.columns.tolist()}')

reg = ExtraTreesRegressor(n_estimators=60, n_jobs=-1, random_state=42)

sfs1 = sfs(reg,
           k_features=3,
           forward=True,
           floating=False,
           verbose=2,
           scoring='r2',
           cv=5)

sfs1.fit(df_fwd_subset, y_train)
```

### Step 3: View Selected Features


```python
print(f'Selected features: {sfs1.k_feature_names_}')

df_fwd_selected = df_fwd_subset[list(sfs1.k_feature_names_)].reset_index(drop=True)
df_fwd_selected.head()
```
