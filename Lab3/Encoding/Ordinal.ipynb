# Encoding Technique 1: Ordinal Encoding

**Dataset:** `Loan_Default.csv`

**When to use:** For **ordinal** (ordered) categorical features. The encoder assigns an integer to each category, preserving the natural rank order.

**Key concept:** The `age` column (`<25`, `25-34`, `35-44`, `45-54`, `55-64`, `>64`) is a perfect example — it has a natural, meaningful order.

---


### Step 1: Setup & Data Loading


```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Load data
df = pd.read_csv('../../data/raw/Loan_Default.csv')
df.drop(['ID', 'year'], axis=1, inplace=True)

print(f'Starting shape: {df.shape}')
df.head()
```

### Step 2: Identify Ordinal Feature

We identify 'age' as the ordinal feature. In a real-world scenario, you would explicitly define the order (e.g., `<25` < `25-34` < ...) to ensure the encoder doesn't just use alphabetical order.


```python
Ordinal_features = ['age']
```

### Step 3: Apply Ordinal Encoding

The `OrdinalEncoder` from scikit-learn converts categorical labels into integer codes.


```python
df_ordinal = df.copy()

enc = OrdinalEncoder()
df_ordinal[Ordinal_features] = enc.fit_transform(df_ordinal[Ordinal_features])
df_ordinal.head()
```

### Step 4: Verify the Result

The `age` column should now contain floats representing the encoded age brackets.


```python
print(df_ordinal['age'].value_counts())
print(f'\nData type of age column: {df_ordinal["age"].dtype}')
```
