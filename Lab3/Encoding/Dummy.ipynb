# Encoding Technique 3: Dummy Encoding

**Dataset:** `Loan_Default.csv`

**When to use:** For **nominal** (unordered) features, as an alternative to One-Hot Encoding that **avoids multicollinearity**.

**Key concept:** Instead of N columns for N categories, Dummy Encoding creates only **N-1 columns**. The dropped category acts as a "reference" (baseline) — it is represented when all other dummy columns are 0.

**Implementation:** `pd.get_dummies(drop_first=True)` — this is a pure-pandas approach, no sklearn needed.

---

### Step 1: Setup, Data Loading & Prep

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

# Load data
df = pd.read_csv('../../data/raw/Loan_Default.csv')
df.drop(['ID', 'year'], axis=1, inplace=True)

categorical_features = df.select_dtypes(include=['object']).columns.tolist()
Ordinal_features = ['age']
Nominal_features = categorical_features.copy()
Nominal_features.remove('age')

# Pre-encode ordinal feature
enc = OrdinalEncoder()
df[Ordinal_features] = enc.fit_transform(df[Ordinal_features])

print(f'Shape before Dummy Encoding: {df.shape}')
df.head()
```

### Step 2: Apply Dummy Encoding

The key parameter is `drop_first=True`. This drops the first category of each feature, creating N-1 columns and preventing multicollinearity.

```python
df_dummy = df.copy()

# Apply dummy encoding — drop_first=True is the key difference from OHE
df_dummy_categorical = pd.get_dummies(data=df_dummy[Nominal_features], drop_first=True)

# Separate numerical features and reassemble
df_dummy_numerical = df_dummy.drop(Nominal_features, axis=1)
df_dummy = pd.concat([df_dummy_numerical, df_dummy_categorical], axis=1)

print(f'Shape after Dummy Encoding: {df_dummy.shape}')
df_dummy.head()
```

### Step 3: Compare Column Counts vs. One-Hot Encoding

For a feature with N categories:
- **One-Hot Encoding** creates **N columns**
- **Dummy Encoding** creates **N-1 columns**

The `Gender` column has 4 categories → OHE creates 4 columns, Dummy creates only 3.

```python
# Show which dummy columns were created for 'Gender'
gender_cols = [c for c in df_dummy.columns if c.startswith('Gender')]
print('Dummy columns for Gender:', gender_cols)
df_dummy[gender_cols].head(10)
```
