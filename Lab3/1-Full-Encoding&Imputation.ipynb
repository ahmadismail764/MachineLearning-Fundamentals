## Encoding Nominal Features by One Hot/Dummy Encoding

### One Hot Encoder

- One-Hot Encoder (OHE)This is the standard approach for nominal data. It creates a new binary column for every unique category in the feature.
- How it works: If a feature has $N$ categories, it creates $N$ columns. A value of 1 indicates the presence of that category, and 0 indicates absence.

```python
loan_data.shape
loan_data2_1 = loan_data.copy()
loan_data2_1.shape
```

```python
OH_encoder = ce.OneHotEncoder(cols=Nominal_features,handle_unknown='return_nan',return_df=True,use_cat_names=True)
```

```python
# Remove categorical columns (will replace with one-hot encoding)
loan_data2_1_numerical_features = loan_data2_1.drop(Nominal_features, axis=1)

# encode categorical features
loan_data2_1_categorical_features = OH_encoder.fit_transform(loan_data2_1[Nominal_features])

# Add one-hot encoded columns to numerical features
loan_data2_1 = pd.concat([loan_data2_1_numerical_features, loan_data2_1_categorical_features], axis=1)
loan_data2_1.columns
```

### One Hot Encoder of SKLearn


```python
loan_data2 = loan_data.copy()
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
OH_encoder = OneHotEncoder(handle_unknown='ignore', sparse_output=False)
OH_cols_train = pd.DataFrame(OH_encoder.fit_transform(loan_data2[Nominal_features]))

# Assign descriptive column names
OH_cols_train.columns = OH_encoder.get_feature_names_out(Nominal_features)

OH_cols_train.index = loan_data2.index

# Remove categorical columns (will replace with one-hot encoding)
num_X_train = loan_data2.drop(Nominal_features, axis=1)

# Add one-hot encoded columns to numerical features
OH_X_train = pd.concat([num_X_train, OH_cols_train], axis=1)
OH_X_train
```

### Dummy Encoder

- Very similar to One-Hot Encoding, but it aims to avoid the "dummy variable trap" (multicollinearity).
- How it works: It creates $N-1$ columns for $N$ categories. The "missing" category is represented when all other columns are 0 (acting as the baseline).


```python
loan_data3 = loan_data.copy()
```

```python
# Remove categorical columns (will replace with one-hot encoding)
loan_data3_numerical_features = loan_data3.drop(Nominal_features, axis=1)

# encode categorical features
loan_data3_categorical_features = pd.get_dummies(data=loan_data3[Nominal_features],drop_first=True)

# Add one-hot encoded columns to numerical features
loan_data3 = pd.concat([loan_data3_numerical_features, loan_data3_categorical_features], axis=1)
loan_data3
```

### Frequency Encoder

- This technique uses the prevalence of a category as its numerical representation.

- How it works: Each category is replaced by the number of times it appears in the dataset (or the percentage/frequency).


```python
loan_data5 = loan_data.copy()
```

```python
for c in (Nominal_features):
    loan_data5[c+'_freq'] = loan_data5[c].map(loan_data5.groupby(c).size() / loan_data5.shape[0])
    indexer = pd.factorize(loan_data5[c], sort=True)[1]
    loan_data5[c] = indexer.get_indexer(loan_data5[c])

loan_data5 =  loan_data5.drop(Nominal_features, axis=1)
loan_data5
```

### Target Encoder

- This is a supervised encoding technique that uses the relationship with the target variable.
- How it works: Each category is replaced by the average value of the target variable for that specific category. For example, in a "Churn" prediction, "New York" would be replaced by the actual churn rate of users in New York.
- Pros: Captures a direct correlation between the feature and the target; handles high cardinality efficiently.


```python
loan_data4 = loan_data.copy()
```

```python
Target_encoder=ce.TargetEncoder(cols=Nominal_features)
```

```python
# Remove categorical columns (will replace with one-hot encoding)
loan_data4_numerical_features = loan_data4.drop(Nominal_features, axis=1)

# encode categorical features
loan_data4_categorical_features = Target_encoder.fit_transform(loan_data4[Nominal_features], loan_data4['loan_amount'])

# Add one-hot encoded columns to numerical features
loan_data4 = pd.concat([loan_data4_numerical_features, loan_data4_categorical_features], axis=1)
loan_data4
```

### Mean/Median Imputer


```python
Cols_to_be_imputed = ['term', 'income', 'age', 'loan_limit_freq', 'approv_in_adv_freq', 'loan_purpose_freq', 'Neg_ammortization_freq', 'submission_of_application_freq']
```
