# Scaling Technique 3: Max Absolute Scaler

**Dataset:** `clean_loan_data.csv` (pre-processed in Lab 3)

**When to use:** When data contains both **positive and negative values** and you want to preserve the sign while scaling to **[-1, 1]**. It is also frequently used for sparse data.

**Formula:** $x_{scaled} = \frac{x}{|x_{max}|}$

---


### Step 1: Setup & Data Loading

We use the cleaned dataset from Lab 3, where encoding and imputation are already completed.


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MaxAbsScaler

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_loan_data.csv')

print(f'Dataset shape: {df.shape}')
df.head()
```

### Step 2: Apply Max Absolute Scaler

The `MaxAbsScaler` divides each feature by its maximum absolute value.


```python
df_maxabs = df.copy()
ma = MaxAbsScaler()

df_maxabs_scaled = pd.DataFrame(ma.fit_transform(df_maxabs), columns=df_maxabs.columns)
df_maxabs_scaled.head()
```

### Step 3: Verify the Result

Check that all absolute values are ≤ 1.


```python
print('Max absolute value per column (should be <= 1):')
print(df_maxabs_scaled.abs().max())
```
