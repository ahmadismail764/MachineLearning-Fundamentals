# Scaling Technique 1: Standardization (Standard Scaler)

**Dataset:** `clean_loan_data.csv` (pre-processed in Lab 3)

**When to use:** When data follows a **Gaussian (Normal) distribution**. This method centers data around 0 with unit variance.

**Formula:** $z = \frac{x - \mu}{\sigma}$

---


### Step 1: Setup & Data Loading

We use the cleaned dataset from Lab 3.


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_loan_data.csv')

print(f'Dataset shape: {df.shape}')
df.head()
```

### Step 2: Apply Standard Scaler

The `StandardScaler` rescales data so it has a **mean of 0** and a **standard deviation of 1**.


```python
df_standard = df.copy()
sc = StandardScaler()

df_standard_scaled = pd.DataFrame(sc.fit_transform(df_standard), columns=df_standard.columns)
df_standard_scaled.head()
```

### Step 3: Verify the Result

Check that the mean is approximately 0 and standard deviation is approximately 1.


```python
print('Mean per column (should be ~0):')
print(df_standard_scaled.mean().round(6))
print('\nStd per column (should be ~1):')
print(df_standard_scaled.std().round(6))
```
