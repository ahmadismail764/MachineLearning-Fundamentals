# Scaling Technique 4: Robust Scaler

**Dataset:** `clean_loan_data.csv` (pre-processed in Lab 3)

**When to use:** When data contains **outliers**. This scaler uses Median and IQR instead of mean/variance, making it robust to extreme values.

**Formula:** $x_{scaled} = \frac{x - Q_2}{Q_3 - Q_1}$

---

### Step 1: Setup & Data Loading

We use the cleaned dataset from Lab 3.

```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import RobustScaler

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_loan_data.csv')

print(f'Dataset shape: {df.shape}')
df.head()
```

### Step 2: Apply Robust Scaler

The `RobustScaler` centers data using the **median** and scales it using the **interquartile range (IQR)**.

```python
df_robust = df.copy()
rs = RobustScaler()

df_robust_scaled = pd.DataFrame(rs.fit_transform(df_robust), columns=df_robust.columns)
df_robust_scaled.head()
```

### Step 3: Verify the Result

The median should be approximately 0 for each column.

```python
print('Median per column (should be ~0):')
print(df_robust_scaled.median().round(6))
```
