# Scaling Technique 2: Min-Max Normalization

**Dataset:** `clean_loan_data.csv` (pre-processed in Lab 3)

**When to use:** When you need to rescale data to a **fixed range [0, 1]**. Note that this method is highly **sensitive to outliers**.

**Formula:** $x_{scaled} = \frac{x - x_{min}}{x_{max} - x_{min}}$

---


### Step 1: Setup & Data Loading

We use the cleaned dataset from Lab 3.


```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_loan_data.csv')

print(f'Dataset shape: {df.shape}')
df.head()
```

### Step 2: Apply Min-Max Scaler

The `MinMaxScaler` rescales each feature to the range **[0, 1]**.


```python
df_minmax = df.copy()
mm = MinMaxScaler()

df_minmax_scaled = pd.DataFrame(mm.fit_transform(df_minmax), columns=df_minmax.columns)
df_minmax_scaled.head()
```

### Step 3: Verify the Result

Check that all values are between 0 and 1.


```python
print('Min per column (should be 0):')
print(df_minmax_scaled.min())
print('\nMax per column (should be 1):')
print(df_minmax_scaled.max())
```
