# Scaling Technique 5: Mean Normalization

**Dataset:** `clean_loan_data.csv` (pre-processed in Lab 3)

**When to use:** Similar to standardization but scales by the **range** instead of the standard deviation.

**Formula:** $x_{scaled} = \frac{x - \text{mean}}{\text{max} - \text{min}}$

> **Note:** Scikit-learn does not have a built-in class for Mean Normalization, so it is implemented manually.

---

### Step 1: Setup & Data Loading

We use the cleaned dataset from Lab 3.

```python
import pandas as pd
import numpy as np

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_loan_data.csv')

print(f'Dataset shape: {df.shape}')
df.head()
```

### Step 2: Apply Mean Normalization (Manual)

We subtract the mean and divide by the range (max - min) for each column.

```python
df_meannorm = df.copy()

for col in df_meannorm.columns:
    col_mean = df_meannorm[col].mean()
    col_max  = df_meannorm[col].max()
    col_min  = df_meannorm[col].min()
    
    # Avoid division by zero if max == min
    if col_max != col_min:
        df_meannorm[col] = (df_meannorm[col] - col_mean) / (col_max - col_min)
    else:
        df_meannorm[col] = 0.0

df_meannorm.head()
```

### Step 3: Verify the Result

The mean should be approximately 0 for each column.

```python
print('Mean per column (should be ~0):')
print(df_meannorm.mean().round(6))
```
