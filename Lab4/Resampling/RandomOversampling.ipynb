# Resampling Technique 1: Random Oversampling

**Dataset:** `clean_insurance_data.csv` (pre-processed in Lab 3)

**When to use:** When the dataset is **imbalanced** (one class has far fewer samples). Randomly **duplicates** minority class samples.

**Key concept:** Increases the number of minority samples by randomly copying existing ones until a desired ratio is reached.

> **Drawback:** Can lead to **overfitting** since it just duplicates existing data points.

---


### Step 1: Setup & Data Loading

We use the cleaned insurance dataset where encoding and deduplication are already done.


```python
import pandas as pd
import numpy as np
from imblearn.over_sampling import RandomOverSampler

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_insurance_data.csv')

y = df['Response']
X = df.drop('Response', axis=1)

print('Class distribution (before resampling):')
print(y.value_counts(normalize=True))
print(f'\nTotal Shape: {df.shape}')
```

### Step 2: Apply Random Oversampling

Using `sampling_strategy=0.5` to make the minority class half the size of the majority class.


```python
oversample = RandomOverSampler(sampling_strategy=0.5)
X_over, y_over = oversample.fit_resample(X, y)

# Reassemble into a DataFrame
df_oversampled = pd.DataFrame(X_over)
df_oversampled['Response'] = y_over

df_oversampled.head()
```

### Step 3: Verify the Result

Check the new class distribution.


```python
print(f'Shape after oversampling: {df_oversampled.shape}')
print('\nClass distribution (after oversampling):')
print(df_oversampled['Response'].value_counts(normalize=True))
```
