# Resampling Technique 3: SMOTE (Synthetic Minority Over-sampling Technique)

**Dataset:** `clean_insurance_data.csv` (pre-processed in Lab 3)

**When to use:** When you want to oversample the minority class **without simply duplicating** rows. Creates **synthetic** examples.

**How it works:** For each minority point, SMOTE finds its **k nearest neighbors** and creates a new synthetic point on the line between them.

> **Advantage over Random Oversampling:** Generates new, unique samples instead of duplicates, reducing overfitting risk.

---

### Step 1: Setup & Data Loading

We use the cleaned insurance dataset.

```python
import pandas as pd
import numpy as np
from imblearn.over_sampling import SMOTE

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_insurance_data.csv')

y = df['Response']
X = df.drop('Response', axis=1)

print('Class distribution (before resampling):')
print(y.value_counts(normalize=True))
print(f'\nTotal Shape: {df.shape}')
```

### Step 2: Apply SMOTE

`sampling_strategy='minority'` will oversample the minority class until it matches the majority class count (50/50 balance).

```python
smote = SMOTE(sampling_strategy='minority')
X_sm, y_sm = smote.fit_resample(X, y)

# Reassemble into a DataFrame
df_smote = pd.DataFrame(X_sm)
df_smote['Response'] = y_sm

df_smote.head()
```

### Step 3: Verify the Result

The classes should now be perfectly balanced.

```python
print(f'Shape after SMOTE: {df_smote.shape}')
print('\nClass distribution (after SMOTE):')
print(df_smote['Response'].value_counts(normalize=True))
```
