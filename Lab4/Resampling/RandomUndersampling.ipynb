# Resampling Technique 2: Random Undersampling

**Dataset:** `clean_insurance_data.csv` (pre-processed in Lab 3)

**When to use:** When the dataset is **imbalanced** and you have **plenty of data**. Randomly **removes** majority class samples.

**Key concept:** Decreases the number of majority samples by randomly discarding them until a desired ratio is reached.

> **Drawback:** Can **lose important information** from the majority class.

---


### Step 1: Setup & Data Loading

We use the cleaned insurance dataset.


```python
import pandas as pd
import numpy as np
from imblearn.under_sampling import RandomUnderSampler

# Load the cleaned dataset
df = pd.read_csv('../../data/Lab4/clean_insurance_data.csv')

y = df['Response']
X = df.drop('Response', axis=1)

print('Class distribution (before resampling):')
print(y.value_counts(normalize=True))
print(f'\nTotal Shape: {df.shape}')
```

### Step 2: Apply Random Undersampling

Using `sampling_strategy=0.5` to reduce the majority class.


```python
undersample = RandomUnderSampler(sampling_strategy=0.5)
X_under, y_under = undersample.fit_resample(X, y)

# Reassemble into a DataFrame
df_undersampled = pd.DataFrame(X_under)
df_undersampled['Response'] = y_under

df_undersampled.head()
```

### Step 3: Verify the Result

Check the new class distribution. The total number of rows should be significantly smaller.


```python
print(f'Shape after undersampling: {df_undersampled.shape}')
print('\nClass distribution (after undersampling):')
print(df_undersampled['Response'].value_counts(normalize=True))
```
